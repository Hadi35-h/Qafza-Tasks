```python
import os
import pandas as pd 
```

```python
# جلب وقراءة الملف من الخخطوة السابقة
df = pd.read_csv("artifacts/joined_FinalTable.csv")
df["order_delivered_customer_date"] = pd.to_datetime(df["order_delivered_customer_date"])       #  تحويل لتواريخ الى  وتخزينها في نفس العمود بدل ان  ننشئ جدول جديد datatime 
df["order_estimated_delivery_date"] = pd.to_datetime(df["order_estimated_delivery_date"])
```

```python
df_delivered = df.dropna(subset=["order_delivered_customer_date"]).copy()  #   استبعاد طلبات لا تحتوي على تاريخ توزيع فعلي وانشاء جدول جديد يحوي على الصفوف التي لها تاريخ وصول فعلي مستقلة عن جدول البيانات الشمامل نسخة جديدة 
df_delivered["is_late"] = (df_delivered["order_delivered_customer_date"] > df_delivered["order_estimated_delivery_date"]).astype(int)   # 0=في الموعد
#1= متأخر 

# اذا كان التاريخ الفعلي  للوصول الطلب اقل من التاريخ المتوقع  فان الطلب غير متأخر  والعكس صحيح 
```

```python
# الليبل spot_check
df_delivered[
[
    "order_id",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "is_late",
]

].head(50)      #    اسطر ال 50 الاولى

# نلاحظ وجود طلبات متأخرة بعد الطلبات العشر الاولى 

```

```python
# جمع الطلبات المتأخرة  وايجاد النسبة بينها وبين الشحنات الغير متأخرة
Count_delv = df_delivered["is_late"].value_counts()  #حساب  عدد الطلبات التي تحمل الفئة 0  الغير متأخرة   والفئات المتأخرة  في تعليمة واحدة 


percentages = df_delivered["is_late"].value_counts(normalize=True) * 100    # حساب النسبة المئوية لكل فئة  من المجموع الكلي 



print(f"******* Result *********")
print(f"(0) In Time non Late : {Count_delv.get(0,0)} row ({percentages.get(0,0): .3f}%)")   # القيمة المقابلة  للفئة 0 الغير متأخرة  اي عددها ومن ثم نسبتها  واذا لم تكن موجودة اعطني 0 
print(f"(1) This is Late : {Count_delv.get(1,0)} row ({percentages.get(1,0): .3f}%)")  # لقيمة المقابلة  للفئة 1  المتأخرة  اي عددها ومن ثم نسبتها  واذا لم تكن موجودة اعطني 0 

# :.3f  عدد عشري بثلاث منازل بعد الفاصلة 
# وكما نلاحظ عدم التوازن في الفئات ظاهر  وهذا امر جيد  ويدل على الدقة 
```
